## Un langage de programmation
Permet de faire une interface entre l'humain et la machine. On écrit des commandes qui sont _interprétées_ de façon linéaire. 
C'est le principe d'un algorithme : une suite finie d'instructions.

### Algorithmique
Pour effectuer une action dans la vraie vie, vous allez devoir effectuer des tâches successives, que l'on peut plus ou moins réduire à une granularité fine. 
Par exemple, pour faire une moyenne sur des notes : 
```   
1- Trouver les notes
2- Les additionner
3- Repérer le nombre de notes dans l'ensemble
4- Diviser la somme des notes par le nombre de notes
```
Dans le cadre d'un langage de programmation, on aurait quelque chose comme :

````python
# Trouver les notes 
notes = [14,12,19,6,17,3]

# Les additionner
total=sum(notes)
# "Imprimer" le total
print(f"Le total est de {total}")

# Repérer le nombre de notes dans l'ensemble
nb=len(notes)
# "Imprimer le nombre de notes"
print(f'il existe {nb} notes différentes')

# Diviser la somme des notes par le nombre de notes
moyenne=total/nb
# "Imprimer la moyenne"
print(f"La moyenne est de {round(moyenne, 2)}")
````

Vous pouvez remarquer au dessus que les différents éléments manipulés ont un nom. Il s'agit des **variables**.    
Par exemple, à la ligne 2, on créé la **variable** `notes` à laquelle on **affecte une valeur** grâce à l'opérateur `=`.         
Python comprend alors que la variable `notes` comprend l'élément `[14,12,19,6,17,3]`, et le stocke en mémoire.    
     
A la ligne `moyenne=total/nb` par exemple, on crée la variable `moyenne` à laquelle on affecte le résultat de la division du contenu de la variable `total` par le contenu de la variable `nb`

Les outils informatiques permettent d'aller plus vite dans certains cas, il s'agit des **librairies** et des **fonctions** associées. 

````python
# On importe la librairie statistics, qui permet de faire du calcul
import statistics

#En faisant appel à la *fonction* mean de la librairie statistics, on obtient : 
moyenne=statistics.mean(notes)
print(f"La moyenne est toujours de {round(moyenne, 2)}")
````

* `dic`, un dictionnaire : 
    - Tout comme les dictionnaires de langues, un dictionnaire en Python associe un élément à un autre (dans un dictionnaire on associe un mot à sa définition).
    Ici, on parlera de clé (_key_) associée à une valeur (_value_).
    Une entrée, c'est à dire `key+value associée`, s'appelle un `item`.

    Un dictionnaire se représente entre accolades `{}`, avec des paires clé-valeurs représentées par `:` et séparées par des virgules :     
    `{ clé 1 : valeur 1 , clé 2 : valeur 2, etc}`    
    Dans un dictionnaire, la clé doit être un élément seul. Il peut s'agir d'un `float`, d'un `str` ou d'un `int`, mais vous ne pourrez par exemple pas mettre une liste en clé.  

    Cependant, la valeur d'une clé peut être de n'importe quelle sorte. Il peut s'agir d'une liste, d'un autre dictionnaire, ou des autres types de données dont nous parlerons juste après. 

    Exemples :
    `{ "bonjour" : 1, "je" : 2, "m'apelle" : 3, "Louis" : 4}` est un dictionnaire.
    `{ "bonjour" : {"type" : "mot", "langue": "français", "position" : 1}, "je" : {"type" : "mot", "langue": "français", "position" : 2 }, "m'appelle" : {"type" : "mot", "langue": "français", "position": 3}, "Louis" : {"type" : "mot", "langue": "français", "position" : 4} }`     
    est un dictionnaire dont chaque valeur est un autre dictionnaire.     
    Pour faciliter la lecture d'un dictionnaire, on peut **indenter** le contenu, c'est à dire représenter les niveaux de hiérarchie grâce aux sauts de lignes et aux espaces : 



````python
dic =   {                                # Début du dictionnaire "dic"  
        "bonjour" :                        # Clé 1
            {                                # Valeur 1 → Début du dictionnaire
                "type" : "mot",                # Clé 1.1 ("type") + valeur 1.1 ("mot")     
                "langue": "français",          # Clé 1.2 ("langue") + valeur 1.2 ("français")
                "position" : 1                 # Clé 1.3 ("position") + valeur 1.3 (1)
            },                               # Fin du dictionnaire → Fin de la valeur 1
        "je" :                             # Clé 2
            {                                # Valeur 2 → Début du dictionnaire
                "type" : "mot",                 # Clé 2.1 ("type") + valeur 2.1 ("mot")     
                "langue": "français",           # Clé 2.2 ("langue") + valeur 2.2 ("français")
                "position" : 2                  # Clé 2.3 ("position") + valeur 2.3 (2)
            },                               # Fin du dictionnaire → Fin de la valeur 2
        "m'appelle" :                      # Clé 3
            {                                # Valeur 3 → Début du dictionnaire
                "type" : "mot",                 # Clé 3.1 ("type") + valeur 3.1 ("mot")   
                "langue": "français",           # Clé 3.2 ("langue") + valeur 3.2 ("français") 
                "position": 3                  # Clé 3.3 ("position") + valeur 3.3 (3)
            },                               # Fin du dictionnaire → Fin de la valeur 2
        "Louis" :                          # Clé 4
            {                                # Valeur 4 → Début du dictionnaire
                "type" : "mot",                 # Clé 4.1 ("type") + valeur 4.1 ("mot") 
                "langue": "français",           # Clé 4.2 ("langue") + valeur 4.2 ("français")  
            "position" : 4                  # Clé 4.3 ("position") + valeur 4.3 (4)
            }
        }

type(dic)
````

## Les boucles, les conditions et l'indentation en Python
Python est un langage qui permet l'utilisation de boucles, et d'exécutions de parties de code selon des conditions données, et se représentent grâce à l'indentation.     

Deux boucles :
* boucle `for` :
Il s'agit de **répéter** une séquence de code de façon **finie**, c'est à dire un nombre x de fois. Le nombre de répétitions est itératif, c'est à dire qu'il correspondra à plusieurs éléments successifs
* boucle `while` : 
Tant que la condition est active, le code continuera de s'éxecuter. Parfait pour créer des boucles infinies ! (n'utilisez pas de boucle while sauf si vous êtes vraiment à l'aise avec Python)


### Exemples de boucle for :

````python
for i in range(10): # range est une fonction qui renvoie de façon incrémentée chaque nombre entre 0 et le nombre affiché-1
    print(i)
````

````python
# On peut aussi itérer sur des éléments d'une variable donnée :
liste=["bonjour", "je", "m'appelle", "Louis"]
for i in liste: # La boucle for ira récupérer chaque élément par son index, et l'affectera à la variable i
    print(i)
````

````python
# Le fonctionnement d'affectation de la variable par l'index du contenu est le même pour tous les éléments avec un index. Pour les strings, il s'agira de la position dans la chaîne de caractères
for i in "abcdef":
    print(i)
````

````python
dictionnaire={"clé_1":"valeur_1", "clé_2":"valeur_2", "clé_3":"valeur_3"}
# Pour un dictionnaire, il s'agira de la clé
for i in dictionnaire:
    print(i)
````

````python
# Pour accéder à la fois aux clés et aux valeurs d'un dictionnaire via la boucle for, on devra créer deux variables, que l'on affectera grâce à la méthode .items():
for i, j in  dictionnaire.items():
    print(f"i : {i}, j: {j}")
````

### Exemple de boucle while :

````python
# On va itérer sur une liste, que l'on réduira petit à petit :
liste = [1, 2, 3, 4, 5]

while len(liste)>0: # Tant que la longueur de la liste (fonction len()) est supérieure à 0 :
    print(liste)    # On imprime la liste
    liste.pop(0)    # On enlève l'élement en première position de la liste → comme la liste est un objet mutable, on n'a pas besoin de le réaffecter. La fonction pop() d'une liste extrait l'élément à l'index donné. 
````

#### Les fonctions if, elif et else 
Permettent de définir des conditions, et se structurent via l'indentation. 
* `if` exprime une condition, si elle est remplie le code du bloc s'exécute.
* `else` permet au programme d'éxecuter du code alternatif, si la condition du bloc if n'est pas remplie. Le test des conditions est ensuite terminé.
* `elif` permet au programme d'éxecuter du code alternatif, si la condition du bloc `if` n'est pas remplie, et d'assurer une continuité des tests de conditions ensuite.    

````python
# Exemple :
# ici, on va tester une condition, à savoir si un nombre est supérieur à 5 (if). S'il n'est pas supérieur à 5, on teste s'il est inférieur à 3 (elif). Dans tous les autres cas, on effectuera le code de else.

for i in range(8): # i aura pour valeur 0, puis 1, etc jusqu'à 6
    if i > 5 :
        print(f"Le nombre {i} est supérieur à 5")
    elif i<3:
        print(f"Le nombre {i} est inférieur à 3")
    else:
        print(f"Le nombre {i} est compris entre 3 et 5") 
````

### Comment arrêter une boucle même si la condition d'origine n'est pas remplie ? 
C'est surtout utile pour les boucles while, mais ça peut l'être pour les boucles for
On peut utiliser trois **instructions**, `break`, `continue` ou `pass`.
* `break` casse la boucle, le programme en sort indépendament du respect de la condition d'origine.
* `continue` revient au début de la boucle en "sautant" l'itération en cours. L'itération actuelle cesse, et on passe à la suivante.
* `pass` permet de gérer la condition en cours (ìf, elif, else) sans que la boucle ne soit affectée. Elle ne gère que la condition au dessus.

## Les paradigmes des langages de programmation
### La Programmation Orientée Objet (POO)
* On distingue plusieurs paradigmes dans les langages de programmation, qui correpondent plus ou moins à la manière d'organiser de concevoir du code.
* Un langage peut supporter plusieurs paradigmes, et c'est le cas de Python.
* Python permet avant tout une Programmation Orientée Objet : 
    * Organisation du code autour d'objets qui représentent des entités :
        *  Ces objets contiennent des attributs (+/- _données_)
        * Ces objets contiennent également des méthodes (+/- _fonctions_)
    * 4 éléments principaux dans la POO :
        * Encapsulation : 
            - On protège les données internes
        * Abstraction :
            - On ne montre que ce qui est utile
        * Héritage :
            - Les différentes hiérarchies d'organisation peuvent hériter de leurs attributs

            - Une même méthode peut amener des comportements différents

### Traduction du concept de POO en code :

````python
class Voiture:
    """
    Classe Voiture
    Cette classe représente les objets voiture.
    Elle encapsule :
    - des données/attributs (prix, consommation, essence)
    - des comportements/fonctions (rouler, mettre de l'essence, calculer l'autonomie)
    """

    def __init__(self, prix, conso_l_100km):
        """
        Constructeur de la classe.
        Il est appelé automatiquement lors de la création d'un objet voiture.
        Les attributs de cette classe sont partagés, et créés à chaque création de nouvel objet voiture

        paramètre 'prix' : prix de la voiture (en euros)
        paramètre 'conso_l_100km' : consommation d'essence en litres pour 100 km
        """

        # Attribut protégé : le prix de la voiture
        # Par convention, le "_" indique qu'il ne doit pas être modifié directement
        self._prix = prix

        # Conversion de la consommation :
        # on passe de "litres / 100 km" à "litres / km"
        self._conso_l_km = conso_l_100km / 100

        # Quantité actuelle d'essence dans le réservoir (en litres)
        # Valeur initiale arbitraire pour l'exemple
        self._essence = 150

    # ------------------------------------------------------------------
    # MÉTHODES D'INFORMATION
    # Ces méthodes ne modifient PAS l'état de l'objet.
    # Elles servent uniquement à obtenir des informations.
    # ------------------------------------------------------------------

    def autonomie(self):
        """
        Calcule et retourne l'autonomie de la voiture en kilomètres.

        Retounr le nombre de kilomètres possibles (float)
        """

        # Si la voiture n'a plus d'essence, elle ne peut pas rouler
        if self._essence <= 0:
            return 0

        # Calcul de l'autonomie :
        # autonomie = quantité d'essence / consommation par km
        autonomie_km = self._essence / self._conso_l_km

        # On arrondit pour un affichage plus propre
        return round(autonomie_km, 2)

    def a_de_l_essence(self):
        """
        Indique si la voiture a encore de l'essence.

        Retourne True si l'essence est > 0, False sinon
        """
        return self._essence > 0

    # ------------------------------------------------------------------
    # MÉTHODES D'ACTION
    # Ces méthodes MODIFIENT l'état de l'objet.
    # ------------------------------------------------------------------

    def mettre_essence(self, litres):
        """
        Ajoute de l'essence dans le réservoir.

        Parmètre litres : quantité d'essence à ajouter (int ou float)
        """

        # On vérifie que la valeur passée est correcte
        if not isinstance(litres, (int, float)):
            raise TypeError("La quantité d'essence doit être un nombre")

        if litres <= 0:
            raise ValueError("La quantité d'essence doit être positive")

        # Modification de l'état interne de l'objet
        self._essence += litres

    def rouler(self, km):
        """
        Fait rouler la voiture sur une certaine distance.

        Paramètre km : distance à parcourir en kilomètres
        """

        # Vérification du type et de la valeur
        if not isinstance(km, (int, float)):
            raise TypeError("La distance doit être un nombre")

        if km <= 0:
            raise ValueError("La distance doit être positive")

        # Calcul de la consommation nécessaire pour cette distance
        consommation = km * self._conso_l_km

        # Vérification de la quantité d'essence disponible
        if consommation > self._essence:
            raise RuntimeError("Pas assez d'essence pour parcourir cette distance")

        # Mise à jour de l'état interne :
        # on retire l'essence consommée
        self._essence -= consommation
````

### Effets de bord et objets mutables

En Python, certaines structures de données sont **mutables** :
- listes
- dictionnaires
- DataFrames ("tableaux" pandas)

Cela signifie que **leur contenu peut être modifié sans créer un nouvel objet**.

Une variable ne contient pas l’objet, mais une **référence** vers cet objet.


````python
liste =[1, 2, 3, 4]
liste.append(5) # La fonction append permet d'ajouter un élémént à la fin d'une liste
print(liste) # Même sans avoir modifié l'objet "liste" en le réaffectant, la liste a changé → c'est un objet mutable
````

### Exemple de référence vers un objet qui entraîne un effet de bord

````python
import pandas as pd 


df= pd.DataFrame({"lieux":["Bourges","Lyon", "Grenoble", "Limoges"],
     "début":[1994,2012,2023,2025],
     "fin":[2012,2023,2025,3000]})
print("Ceci est notre premier DataFrame")
print(df.head(4))
print("---------------------------")

# On déclare une variable df_2, que l'on modifiera ensuite, équivalente au premier df construit
df_2=df

# On modifie df_2, sans modifier df
df_2['lieux']=[1,2,3,4]

# On affiche les deux dataframes → les deux sont directement modifiés !
print("df_2")
print(df_2)
print("---------------------------")
print("df d'origine après modification de df_2")
print(df)
````

**Mais alors, pourquoi ça modifie les deux ?**    
Écrire ``df_2=df`` équivaut à associer df_2 et df au **même objet** !     
df_2 est simplement un alias de df, ils ont la même référence.


Pandas est un outil **impératif** et les dataframe qu'il produit sont **mutables** → attention aux effets de bord !

### Programmation procédurale 
En programmation procédurale, l'indée c'est de découper son code en blocs que l'on va appeler à plusieurs endroits. On crée des **fonctions**.    
Pour créer une fonction en Python, on la définit via l'utilisation de `def`, et on explique les différentes étapes à suivre à partir des `arguments` que l'on passe à la fonction.
Une fonction s'éxecute de façon locale et non globale. L'état de la mémoire ne change pas lors de son éxecution. Elle **retourne** une valeur lorsque l'on s'en sert.

````python
def multiplication (a, b): # Ici on donne un nom à la fonction (multiplication) et on lui donne également des arguments (a et b). Attention, la ligne doit se terminer par :
    result=a*b             # Ici on explique de façon impérative les différentes manipulations que l'ordinateur devra faire sur les données lors de la fonction. Dans ce cas, on crée une variable result à laquelle on affecte la valeur a multiplié par b
    return result          # Une fonction, par principe, s'utilise dans d'autres parties du code. On dit qu'on "l'appelle". Elle renvoie un élément à la fin de ses calculs. Pour renvoyer cet élément, on utilise return
````

````python
print(multiplication(3,4))
print("La fonction multiplication, appelée avec les arguments 3 et 4, renvoie 12.")
````

**Si on reprend l'exemple du DataFrame qui se modifie :**

````python
import pandas as pd 

df= pd.DataFrame({"lieux":["Bourges","Lyon", "Grenoble", "Limoges"],
     "début":[1994,2012,2023,2025],
     "fin":[2012,2023,2025,3000]})
print("Ceci est notre premier DataFrame")
print(df)
print("---------------------------")


def df_change(df):
    # On créé une copie de df que l'on affecte à une variable df_2
    df_2=df.copy()
    # On modifie df_2, sans modifier df_1
    df_2['lieux']=[1,2,3,4]
    # On retourne df_2, que l'on vient de modifier
    return df_2

df_modifié=df_change(df)

# On affiche les deux dataframes → les deux sont directement modifiés !
print("df_modifié")
print(df_modifié)
print("---------------------------")
print("df d'origine")
print(df)
    
````
