## Un langage de programmation
Permet de faire une interface entre l'humain et la machine. On écrit des commandes qui sont _interprétées_ de façon linéaire. 
C'est le principe d'un algorithme : une suite finie d'instructions.

### Algorithmique
Pour effectuer une action dans la vraie vie, vous allez devoir effectuer des tâches successives, que l'on peut plus ou moins réduire à une granularité fine. 
Par exemple, pour faire une moyenne sur des notes : 
```   
1- Trouver les notes
2- Les additionner
3- Repérer le nombre de notes dans l'ensemble
4- Diviser la somme des notes par le nombre de notes
```
Dans le cadre d'un langage de programmation, on aurait quelque chose comme :

````python
# Trouver les notes 
notes = [14,12,19,6,17,3]

# Les additionner
total=sum(notes)
# "Imprimer" le total
print(f"Le total est de {total}")

# Repérer le nombre de notes dans l'ensemble
nb=len(notes)
# "Imprimer le nombre de notes"
print(f'il existe {nb} notes différentes')

# Diviser la somme des notes par le nombre de notes
moyenne=total/nb
# "Imprimer la moyenne"
print(f"La moyenne est de {round(moyenne, 2)}")
````

Vous pouvez remarquer au dessus que les différents éléments manipulés ont un nom. Il s'agit des **variables**.
Par exemple, à la ligne 2, on créé la **variable** `notes` à laquelle on associe une donnée grâce au `=`.     
Python comprend alors que la variable `notes` comprend l'élément `[14,12,19,6,17,3]`, et le stocke en mémoire.

A la ligne `moyenne=total/nb` par exemple, on crée la variable `moyenne` grâce à la division du contenu de la variable `total` par le contenu de la variable `nb`

Les outils informatiques permettent d'aller plus vite dans certains cas, il s'agit des **librairies** et des **fonctions** associées. 

````python
# On importe la librairie statistics, qui permet de faire du calcul
import statistics

#En faisant appel à la *fonction* mean de la librairie statistics, on obtient : 
moyenne=statistics.mean(notes)
print(f"La moyenne est toujours de {round(moyenne, 2)}")
````

## Les Types de données en Python

Vous l'avez sûrement déjà remarqué, les différents éléments écrits depuis le début le sont de manières différentes. Chaque **façon** de représenter une information lui octroie de facto un type.    
**Types de "base" :**
- `int`, les intégraux :
    * Ce sont des nombres entiers. Ils se représentent par un ou plusieurs chiffres, sans guillemets. Le chiffre le plus à gauche ne doit pas être un 0.    
    Exemples : `1`, `4733842`.

- `float`, les décimaux :
    * Ce sont des nombres décimaux. Ils se représentent par un ou plusieurs chiffres, sans guillemets. La virgule est représentée par un point.    
    Exemples : `1.42`, `0.4733842`.

````python
decimal=0.4733842
type(decimal)
````

* `str`, les chaînes de caractères, ou _strings_:
    - Ce sont des chaînes de caractères alphanumériques. Ils sont entourés de guillemets doubles (touche 3 du clavier) `"`, ou de guillemets simples (touche 4 du clavier) `'`. Ils seront considérés comme du texte.     
    Exemples : `"Bonjour je suis une chaîne de caractères"`, `"42"`

````python
string="Bonjour je suis une chaîne de caractères"
type(string)
````

**Ces différentes données peuvent être regroupés dans plusieurs éléments**
* `list`, une liste d'éléments:
    - Il s'agit d'une manière de grouper plusieurs éléments de façon séquentielle. Les différents éléments d'une liste n'ont pas besoin d'avoir le même type.    
    Une liste est entourée de crochets, et chaque élément est séparé par une virgule.   
    Exemple : `[1, "bonjour", 0.42]` est une liste. 

    Chaque élément d'une liste possède une position, ou index. Elle permet de le repérer au sein de la liste, et d'y accéder. La numérotation des indexs en python commence à 0, et est incrémentée de 1 à chaque élément.
    Dans l'exemple ci dessus, 1 est à l'index 0, "bonjour" à l'index 1, 0.42 à l'index 2.

    Pour **accéder** à un élément d'une liste, on écrit le nom de la variable de type liste, suivi de l'index de l'élément que l'on veut atteindre entre crochets

````python
liste=[1, "bonjour", 0.42]
type(liste[0])
````

* `dic`, un dictionnaire : 
    - Tout comme les dictionnaires de langues, un dictionnaire en Python associe un élément à un autre (dans un dictionnaire on associe un mot à sa définition).
    Ici, on parlera de clé (_key_) associée à une valeur (_value_).
    Une entrée, c'est à dire `key+value associée`, s'appelle un `item`.

    Un dictionnaire se représente entre accolades `{}`, avec des paires clé-valeurs représentées par `:` et séparées par des virgules :     
    `{ clé 1 : valeur 1 , clé 2 : valeur 2, etc}`    
    Dans un dictionnaire, la clé doit être un élément seul. Il peut s'agir d'un `float`, d'un `str` ou d'un `int`, mais vous ne pourrez par exemple pas mettre une liste en clé.  

    Cependant, la valeur d'une clé peut être de n'importe quelle sorte. Il peut s'agir d'une liste, d'un autre dictionnaire, ou des autres types de données dont nous parlerons juste après. 

    Exemples :
    `{ "bonjour" : 1, "je" : 2, "m'apelle" : 3, "Louis" : 4}` est un dictionnaire.
    `{ "bonjour" : {"type" : "mot", "langue": "français", "position" : 1}, "je" : {"type" : "mot", "langue": "français", "position" : 2 }, "m'appelle" : {"type" : "mot", "langue": "français", "position": 3}, "Louis" : {"type" : "mot", "langue": "français", "position" : 4} }`     
    est un dictionnaire dont chaque valeur est un autre dictionnaire.     
    Pour faciliter la lecture d'un dictionnaire, on peut **indenter** le contenu, c'est à dire représenter les niveaux de hiérarchie grâce aux sauts de lignes et aux espaces : 



````python
dic =   {                                # Début du dictionnaire "dic"  
        "bonjour" :                        # Clé 1
            {                                # Valeur 1 → Début du dictionnaire
                "type" : "mot",                # Clé 1.1 ("type") + valeur 1.1 ("mot")     
                "langue": "français",          # Clé 1.2 ("langue") + valeur 1.2 ("français")
                "position" : 1                 # Clé 1.3 ("position") + valeur 1.3 (1)
            },                               # Fin du dictionnaire → Fin de la valeur 1
        "je" :                             # Clé 2
            {                                # Valeur 2 → Début du dictionnaire
                "type" : "mot",                 # Clé 2.1 ("type") + valeur 2.1 ("mot")     
                "langue": "français",           # Clé 2.2 ("langue") + valeur 2.2 ("français")
                "position" : 2                  # Clé 2.3 ("position") + valeur 2.3 (2)
            },                               # Fin du dictionnaire → Fin de la valeur 2
        "m'appelle" :                      # Clé 3
            {                                # Valeur 3 → Début du dictionnaire
                "type" : "mot",                 # Clé 3.1 ("type") + valeur 3.1 ("mot")   
                "langue": "français",           # Clé 3.2 ("langue") + valeur 3.2 ("français") 
                "position": 3                  # Clé 3.3 ("position") + valeur 3.3 (3)
            },                               # Fin du dictionnaire → Fin de la valeur 2
        "Louis" :                          # Clé 4
            {                                # Valeur 4 → Début du dictionnaire
                "type" : "mot",                 # Clé 4.1 ("type") + valeur 4.1 ("mot") 
                "langue": "français",           # Clé 4.2 ("langue") + valeur 4.2 ("français")  
            "position" : 4                  # Clé 4.3 ("position") + valeur 4.3 (4)
            }
        }

type(dic)
````

Pour accéder aux différents éléments d'un dictionnaire, on passe par les clés, dans la même syntaxe que les index des listes :
Pour récupérer la valeur associée à la clé "bonjour" du dictionnaire d'au dessus, on tapera le nom de la variable suivi du nom de la clé entre crochets.

````python
dic["bonjour"]
````

Ici, on a bien la valeur associée, à savoir le dictionnaire. Si l'on veut accéder aux valeurs de ce dictionnaire, on met le nom de la clé à atteindre à droite de l'élément précédent, puisqu'il nous donne le dictionnaire.

````python
dic["bonjour"]["position"]
````

## Traduction du concept de POO en code :

````python
class Voiture:
    """
    Classe Voiture
    Cette classe représente les objets voiture.
    Elle encapsule :
    - des données/attributs (prix, consommation, essence)
    - des comportements/fonctions (rouler, mettre de l'essence, calculer l'autonomie)
    """

    def __init__(self, prix, conso_l_100km):
        """
        Constructeur de la classe.
        Il est appelé automatiquement lors de la création d'un objet voiture.
        Les attributs de cette classe sont partagés, et créés à chaque création de nouvel objet voiture

        paramètre 'prix' : prix de la voiture (en euros)
        paramètre 'conso_l_100km' : consommation d'essence en litres pour 100 km
        """

        # Attribut protégé : le prix de la voiture
        # Par convention, le "_" indique qu'il ne doit pas être modifié directement
        self._prix = prix

        # Conversion de la consommation :
        # on passe de "litres / 100 km" à "litres / km"
        self._conso_l_km = conso_l_100km / 100

        # Quantité actuelle d'essence dans le réservoir (en litres)
        # Valeur initiale arbitraire pour l'exemple
        self._essence = 150

    # ------------------------------------------------------------------
    # MÉTHODES D'INFORMATION
    # Ces méthodes ne modifient PAS l'état de l'objet.
    # Elles servent uniquement à obtenir des informations.
    # ------------------------------------------------------------------

    def autonomie(self):
        """
        Calcule et retourne l'autonomie de la voiture en kilomètres.

        Retounr le nombre de kilomètres possibles (float)
        """

        # Si la voiture n'a plus d'essence, elle ne peut pas rouler
        if self._essence <= 0:
            return 0

        # Calcul de l'autonomie :
        # autonomie = quantité d'essence / consommation par km
        autonomie_km = self._essence / self._conso_l_km

        # On arrondit pour un affichage plus propre
        return round(autonomie_km, 2)

    def a_de_l_essence(self):
        """
        Indique si la voiture a encore de l'essence.

        Retourne True si l'essence est > 0, False sinon
        """
        return self._essence > 0

    # ------------------------------------------------------------------
    # MÉTHODES D'ACTION
    # Ces méthodes MODIFIENT l'état de l'objet.
    # ------------------------------------------------------------------

    def mettre_essence(self, litres):
        """
        Ajoute de l'essence dans le réservoir.

        Parmètre litres : quantité d'essence à ajouter (int ou float)
        """

        # On vérifie que la valeur passée est correcte
        if not isinstance(litres, (int, float)):
            raise TypeError("La quantité d'essence doit être un nombre")

        if litres <= 0:
            raise ValueError("La quantité d'essence doit être positive")

        # Modification de l'état interne de l'objet
        self._essence += litres

    def rouler(self, km):
        """
        Fait rouler la voiture sur une certaine distance.

        Paramètre km : distance à parcourir en kilomètres
        """

        # Vérification du type et de la valeur
        if not isinstance(km, (int, float)):
            raise TypeError("La distance doit être un nombre")

        if km <= 0:
            raise ValueError("La distance doit être positive")

        # Calcul de la consommation nécessaire pour cette distance
        consommation = km * self._conso_l_km

        # Vérification de la quantité d'essence disponible
        if consommation > self._essence:
            raise RuntimeError("Pas assez d'essence pour parcourir cette distance")

        # Mise à jour de l'état interne :
        # on retire l'essence consommée
        self._essence -= consommation
````
