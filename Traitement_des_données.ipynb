## Un langage de programmation
Permet de faire une interface entre l'humain et la machine. On écrit des commandes qui sont _interprétées_ de façon linéaire. 
C'est le principe d'un algorithme : une suite finie d'instructions.

### Algorithmique
Pour effectuer une action dans la vraie vie, vous allez devoir effectuer des tâches successives, que l'on peut plus ou moins réduire à une granularité fine. 
Par exemple, pour faire une moyenne sur des notes : 
```   
1- Trouver les notes
2- Les additionner
3- Repérer le nombre de notes dans l'ensemble
4- Diviser la somme des notes par le nombre de notes
```
Dans le cadre d'un langage de programmation, on aurait quelque chose comme :

````python
#Trouver les notes 
notes = [14,12,19,6,17,3]

#Les additionner
total=sum(notes)
#"Imprimer" le total
print(f"Le total est de {total}")

#repérer le nombre de notes dans l'ensemble
nb=len(notes)
#"Imprimer le nombre de notes"
print(f'il existe {nb} notes différentes')

#Diviser la somme des notes par le nombre de notes
moyenne=total/nb
#"Imprimer la moyenne"
print(f"La moyenne est de {round(moyenne, 2)}")
````

Les outils informatiques permettent d'aller plus vite dans certains cas, il s'agit des **librairies** et des **fonctions** associées. 

````python
# On importe la librairie statistics, qui permet de faire du calcul
import statistics

#En faisant appel à la *fonction* mean de la librairie statistics, on obtient : 
moyenne=statistics.mean(notes)
print(f"La moyenne est toujours de {round(moyenne, 2)}")
````

## Spécificités de Python 
* On distingue plusieurs paradigmes dans les langages de programmation, qui correpondent plus ou moins à la manière d'organiser de concevoir du code.
* Un langage peut supporter plusieurs paradigmes, et c'est le cas de Python.
* Python permet avant tout une Programmation Orientée Objet : 
    * Organisation du code autour d'objets qui représentent des entités :
        *  Ces objets contiennent des attributs (+/- _données_)
        * Ces objets contiennent également des méthodes (+/- _fonctions_)
    * 4 éléments principaux dans la POO :
        * Encapsulation : 
            - On protège les données internes
        * Abstraction :
            - On ne montre que ce qui est utile
        * Héritage :
            - Les différentes hiérarchies d'organisation peuvent hériter de leurs attributs
        * Polymorphisme :
            - Une même méthode peut amener des comportements différents
    * Quel intérêt pour la POO ?
        * Applications complexes (ex : QGIS tourne en python)
        * Interfaces graphiques
        * Jeux 
        * Frameworks (ex : Deezer et Spotify tournent en Python, via Django)
    * Quand ne pas utiliser la POO ? 
        * Scripts simples 
        * Calcul

## Traduction du concept de POO en code :

````python
class Voiture: # Définition d'un objet via la création d'une classe
    def __init__(self, prix, essencekm): # Au moment de la création d'un objet de la classe Voiture, on devra spécifier un prix et la consommation d'essence aux 100km
        self._prix = prix # Le prix sera alors attribué à cette élément de la classe. 
        self._essencekm = essencekm/100
        self.essence = 150

    def essence(self):
        if self.essence == 0:
            print("Mettre de l'essence !")
        else:
            print("Vous pouvez encore faire {self.essencekm/100}")

    def mettre_essence(self, ajout):
        if isinstance(ajout, int):
            self.niveau_essence+=ajout
        else:
            print(f"Comment voulez vous ajouter {ajout}L d'essence à une voiture ?!")
    
    def rouler(self, km):
        if isinstance(km,int):
            self.niveau_essence=self.niveau_essence-(self.essencekm*km)
        else:
            print(f"Comment voulez vous rouler {km}km ?")

voiture = Voiture(20000, 6)
voiture.rouler(70)
print(voiture.essence())
````
